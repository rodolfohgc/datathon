## Limpar espaços extras

### Padronizar a coluna 'Fase'

Para garantir consistência nas análises futuras, padronizaremos os valores na coluna `Fase`. O objetivo é unificar diferentes formatos (e.g., '8E', '7', 'FASE 1') para um padrão 'FASE X', onde 'X' é o número da fase. A fase 'ALFA' será mantida como está.

```python
import re

# Certificar que a coluna 'Fase' é do tipo string para aplicar regex e métodos de string
df_long['Fase'] = df_long['Fase'].astype(str)

def standardize_fase(fase_str):
    if pd.isna(fase_str):
        return pd.NA

    fase_str = fase_str.strip().upper() # Remover espaços em branco e converter para maiúsculas

    # Manter 'ALFA' como está
    if fase_str == 'ALFA':
        return 'ALFA'

    # Padronizar formatos como 'FASE X', 'FASE  X', 'FASEX'
    match_fase_num = re.match(r'FASE\s*(\d+)', fase_str)
    if match_fase_num:
        return f'FASE {match_fase_num.group(1)}'

    # Padronizar formatos como 'X', 'XA', 'XB', 'XN' (onde X é um dígito)
    match_num_alpha = re.match(r'(\d+)[A-Z]*', fase_str)
    if match_num_alpha:
        return f'FASE {match_num_alpha.group(1)}'

    return fase_str # Retorna o valor original se não houver correspondência

df_long['Fase'] = df_long['Fase'].apply(standardize_fase)

print("Valores únicos da coluna 'Fase' após padronização:")
print(df_long['Fase'].unique())
```

```python
# Limpa espaços extras para que 'Escola Pública' e ' Escola Pública ' sejam agrupados juntos na análise
cols_categoricas = ['Instituição de ensino', 'Pedra', 'Indicado', 'Atingiu PV']
for col in cols_categoricas:
    if col in df_long.columns:
        df_long[col] = df_long[col].astype(str).str.strip().str.title()
        # Volta valores NaN que viraram string 'Nan' pelo cast acima
        df_long[col] = df_long[col].replace('Nan', pd.NA)
```

Para analisar a defasagem dos alunos, categorizaremos o indicador IAN (Índice de Adequação ao Nível) em três níveis:

*   **Moderadamente Defasado**: 5.0 <= IAN < 7.5
*   **Adequado**: IAN >= 7.5

Vamos aplicar essa categorização e contar o número de alunos em cada grupo por ano de pesquisa.

```python
# Definir as categorias de defasagem
def categorize_ian(ian):
    if ian < 5.0:
        return 'Severamente Defasado'
    elif 5.0 <= ian < 7.5:
        return 'Moderadamente Defasado'
    else:
        return 'Adequado'

df_long['IAN_Categoria'] = df_long['IAN'].apply(categorize_ian)

# Contar alunos por categoria e ano
defasagem_por_ano = df_long.groupby(['Ano Pesquisa', 'IAN_Categoria']).size().unstack(fill_value=0)

# Exibir a tabela de contagem
display(defasagem_por_ano)

# Calcular a proporção de cada categoria por ano
defasagem_proporcao_por_ano = defasagem_por_ano.apply(lambda x: x / x.sum(), axis=1)

# Exibir a tabela de proporção
print("\nProporção de alunos por categoria de IAN por Ano de Pesquisa:")
display(defasagem_proporcao_por_ano)
```

## Análise da Evolução Individual do IAN por Ano de Ingresso

Para entender a trajetória dos alunos, vamos focar na evolução do IAN para cada estudante ('RA') ao longo dos anos de pesquisa, agrupando pelo 'Ano ingresso'. Isso nos permitirá identificar se os alunos avançaram, mantiveram ou decaíram em seu nível de adequação.

```python
# Selecionar colunas relevantes e ordenar para facilitar a análise de evolução
df_evolution = df_long[['RA', 'Ano ingresso', 'Ano Pesquisa', 'IAN']].copy()
df_evolution.sort_values(by=['RA', 'Ano Pesquisa'], inplace=True)

# Calcular a mudança no IAN entre anos consecutivos para cada aluno
df_evolution['IAN_anterior'] = df_evolution.groupby('RA')['IAN'].shift(1)
df_evolution['Mudanca_IAN'] = df_evolution['IAN'] - df_evolution['IAN_anterior']

# Remover as primeiras entradas de cada aluno que terão IAN_anterior como NaN
df_evolution.dropna(subset=['Mudanca_IAN'], inplace=True)

# Categorizar a mudança no IAN
def categorize_ian_change(change):
    if change > 0:
        return 'Avançou'
    elif change < 0:
        return 'Decaiu'
    else:
        return 'Manteve'

df_evolution['Status_IAN'] = df_evolution['Mudanca_IAN'].apply(categorize_ian_change)

# Exibir as primeiras linhas com as novas colunas
print("Evolução do IAN por Aluno e Ano de Pesquisa:")
display(df_evolution.head())
```

### Análise de Transição de IAN entre Anos (2022-2023 e 2023-2024)

Vamos visualizar a movimentação dos alunos entre as categorias de IAN (Severamente Defasado, Moderadamente Defasado, Adequado) de um ano para o outro usando um Gráfico de Sankey. Isso nos ajudará a entender a dinâmica de progresso e regressão no nível de adequação.


```python
import plotly.graph_objects as go

# Preparar os dados para análise de transição
df_transition = df_long[['RA', 'Ano Pesquisa', 'IAN_Categoria']].copy()
df_transition.sort_values(by=['RA', 'Ano Pesquisa'], inplace=True)

# Criar colunas para a categoria IAN do ano atual e do ano seguinte
df_transition['IAN_Categoria_Ano_Seguinte'] = df_transition.groupby('RA')['IAN_Categoria'].shift(-1)

# Remover linhas onde não há dados para o ano seguinte (último ano de cada aluno)
df_transition.dropna(subset=['IAN_Categoria', 'IAN_Categoria_Ano_Seguinte'], inplace=True)

# Filtrar para transições de 2022 para 2023 e de 2023 para 2024
df_2022_2023 = df_transition[(df_transition['Ano Pesquisa'] == 2022) & (df_transition['Ano Pesquisa'] + 1 == 2023)]
df_2023_2024 = df_transition[(df_transition['Ano Pesquisa'] == 2023) & (df_transition['Ano Pesquisa'] + 1 == 2024)]

def create_sankey_data(df_year_transition, year_from, year_to):
    # Contar as transições
    transitions = df_year_transition.groupby(['IAN_Categoria', 'IAN_Categoria_Ano_Seguinte']).size().reset_index(name='Count')

    # Mapear categorias para nós do Sankey
    unique_categories = sorted(df_year_transition['IAN_Categoria'].unique().tolist() + df_year_transition['IAN_Categoria_Ano_Seguinte'].unique().tolist())

    # Adicionar prefixos de ano para diferenciar os nós no Sankey
    labels = [f'{year_from} - {cat}' for cat in unique_categories] + \
             [f'{year_to} - {cat}' for cat in unique_categories]

    # Criar um mapa de índice para os nós
    node_map = {label: i for i, label in enumerate(labels)}

    sources = []
    targets = []
    values = []

    for _, row in transitions.iterrows():
        source_label = f'{year_from} - {row["IAN_Categoria"]}'
        target_label = f'{year_to} - {row["IAN_Categoria_Ano_Seguinte"]}'

        if source_label in node_map and target_label in node_map:
            sources.append(node_map[source_label])
            targets.append(node_map[target_label])
            values.append(row['Count'])

    return {'node_labels': labels, 'sources': sources, 'targets': targets, 'values': values}

# Gerar dados para o Sankey 2022 -> 2023
sankey_data_22_23 = create_sankey_data(df_2022_2023, 2022, 2023)

# Gerar dados para o Sankey 2023 -> 2024
sankey_data_23_24 = create_sankey_data(df_2023_2024, 2023, 2024)

# Plotar Sankey 2022 -> 2023
fig_22_23 = go.Figure(data=[go.Sankey(
    node=dict(
        pad=15,
        thickness=20,
        line=dict(color="black", width=0.5),
        label=sankey_data_22_23['node_labels']
    ),
    link=dict(
        source=sankey_data_22_23['sources'],
        target=sankey_data_22_23['targets'],
        value=sankey_data_22_23['values']
    ))
])
fig_22_23.update_layout(title_text="Transição das Categorias de IAN (2022 para 2023)", font_size=10)
fig_22_23.show()

# Plotar Sankey 2023 -> 2024
fig_23_24 = go.Figure(data=[go.Sankey(
    node=dict(
        pad=15,
        thickness=20,
        line=dict(color="black", width=0.5),
        label=sankey_data_23_24['node_labels']
    ),
    link=dict(
        source=sankey_data_23_24['sources'],
        target=sankey_data_23_24['targets'],
        value=sankey_data_23_24['values']
    ))
])
fig_23_24.update_layout(title_text="Transição das Categorias de IAN (2023 para 2024)", font_size=10)
fig_23_24.show()

# Calcular quantos "Severamente Defasados" em 2022 pularam para "Adequado" em 2023
severely_defasado_2022_to_adequado_2023 = df_2022_2023[
    (df_2022_2023['IAN_Categoria'] == 'Severamente Defasado') &
    (df_2022_2023['IAN_Categoria_Ano_Seguinte'] == 'Adequado')
].shape[0]

print(f"\nNúmero de alunos 'Severamente Defasados' em 2022 que se tornaram 'Adequados' em 2023: {severely_defasado_2022_to_adequado_2023}")
```

Para analisar o desempenho acadêmico dos alunos, categorizaremos o indicador IDA (Indicador de Desempenho Acadêmico) em três níveis, usando a escala de 0 a 10:

*   **Baixo Desempenho**: IDA < 5.0
*   **Médio Desempenho**: 5.0 <= IDA < 7.5
*   **Alto Desempenho**: IDA >= 7.5

Vamos aplicar essa categorização e contar o número de alunos em cada grupo por ano de pesquisa.

```python
# Definir as categorias de desempenho para IDA
def categorize_ida(ida):
    if ida < 5.0:
        return 'Baixo Desempenho'
    elif 5.0 <= ida < 7.5:
        return 'Médio Desempenho'
    else:
        return 'Alto Desempenho'

df_long['IDA_Categoria'] = df_long['IDA'].apply(categorize_ida)

# Contar alunos por categoria e ano
desempenho_por_ano_ida = df_long.groupby(['Ano Pesquisa', 'IDA_Categoria']).size().unstack(fill_value=0)

# Exibir a tabela de contagem
print("Contagem de alunos por categoria de IDA por Ano de Pesquisa:")
display(desempenho_por_ano_ida)

# Calcular a proporção de cada categoria por ano
desempenho_proporcao_por_ano_ida = desempenho_por_ano_ida.apply(lambda x: x / x.sum(), axis=1)

# Exibir a tabela de proporção
print("\nProporção de alunos por categoria de IDA por Ano de Pesquisa:")
display(desempenho_proporcao_por_ano_ida)
```

## Análise da Evolução Individual do IDA por Ano de Ingresso

Para entender a trajetória individual dos alunos, vamos focar na evolução do IDA para cada estudante ('RA') ao longo dos anos de pesquisa, agrupando pelo 'Ano ingresso'. Isso nos permitirá identificar se os alunos avançaram, mantiveram ou decaíram em seu desempenho acadêmico.

```python
# Selecionar colunas relevantes e ordenar para facilitar a análise de evolução
df_evolution_ida = df_long[['RA', 'Ano ingresso', 'Ano Pesquisa', 'IDA']].copy()
df_evolution_ida.sort_values(by=['RA', 'Ano Pesquisa'], inplace=True)

# Calcular a mudança no IDA entre anos consecutivos para cada aluno
df_evolution_ida['IDA_anterior'] = df_evolution_ida.groupby('RA')['IDA'].shift(1)
df_evolution_ida['Mudanca_IDA'] = df_evolution_ida['IDA'] - df_evolution_ida['IDA_anterior']

# Remover as primeiras entradas de cada aluno que terão IDA_anterior como NaN
df_evolution_ida.dropna(subset=['Mudanca_IDA'], inplace=True)

# Categorizar a mudança no IDA
def categorize_ida_change(change):
    if change > 0:
        return 'Avançou'
    elif change < 0:
        return 'Decaiu'
    else:
        return 'Manteve'

df_evolution_ida['Status_IDA'] = df_evolution_ida['Mudanca_IDA'].apply(categorize_ida_change)

# Exibir as primeiras linhas com as novas colunas
print("Evolução do IDA por Aluno e Ano de Pesquisa:")
display(df_evolution_ida.head())
```

### Resumo da Evolução do IDA por Ano de Ingresso e Ano de Pesquisa

Vamos analisar a quantidade e proporção de alunos que avançaram, decaíram ou mantiveram seu nível de IDA, segmentado pelo ano de ingresso e pelo ano da pesquisa.

```python
# Contar o número de alunos por status de IDA, ano de ingresso e ano de pesquisa
summary_ida_evolution = df_evolution_ida.groupby(['Ano ingresso', 'Ano Pesquisa', 'Status_IDA']).size().unstack(fill_value=0)

# Calcular a proporção
proportion_ida_evolution = summary_ida_evolution.div(summary_ida_evolution.sum(axis=1), axis=0)

print("Contagem de alunos por Status do IDA, Ano de Ingresso e Ano de Pesquisa:")
display(summary_ida_evolution)

print("\nProporção de alunos por Status do IDA, Ano de Ingresso e Ano de Pesquisa:")
display(proportion_ida_evolution)
```

```python
# Definir as categorias de engajamento para IEG
def categorize_ieg(ieg):
    if pd.isna(ieg):
        return pd.NA
    if ieg < 5.0:
        return 'Baixo Engajamento'
    elif 5.0 <= ieg < 7.5:
        return 'Médio Engajamento'
    else:
        return 'Alto Engajamento'

df_long['IEG_Categoria'] = df_long['IEG'].apply(categorize_ieg)

# Contar alunos por categoria e ano
engajamento_por_ano_ieg = df_long.groupby(['Ano Pesquisa', 'IEG_Categoria']).size().unstack(fill_value=0)

# Exibir a tabela de contagem
print("Contagem de alunos por categoria de IEG por Ano de Pesquisa:")
display(engajamento_por_ano_ieg)

# Calcular a proporção de cada categoria por ano
engajamento_proporcao_por_ano_ieg = engajamento_por_ano_ieg.apply(lambda x: x / x.sum(), axis=1)

# Exibir a tabela de proporção
print("\nProporção de alunos por categoria de IEG por Ano de Pesquisa:")
display(engajamento_proporcao_por_ano_ieg)
```

### Causalidade Defasada no Tempo: $\Delta$ IEG (Ano N para N+1) vs. $\Delta$ IDA (Ano N+1 para N+2)

Para verificar se o aumento de engajamento antecede a melhora de nota, vamos calcular a correlação entre a variação do IEG de um ano para o próximo e a variação do IDA no ano seguinte. Especificamente, buscaremos a correlação entre:

*   **$\Delta$ IEG (2022 para 2023)**: Mudança no Engajamento de 2022 para 2023.
*   **$\Delta$ IDA (2023 para 2024)**: Mudança no Desempenho Acadêmico de 2023 para 2024.

Isso nos ajudará a entender se o engajamento em um período pode predizer o desempenho acadêmico futuro.

```python
import pandas as pd

# Preparar o DataFrame para calcular as diferenças ano a ano
df_causalidade = df_long[['RA', 'Ano Pesquisa', 'IEG', 'IDA']].copy()
df_causalidade.sort_values(by=['RA', 'Ano Pesquisa'], inplace=True)

# Pivotar o DataFrame para ter IEG e IDA para cada RA por ano como colunas
df_pivot = df_causalidade.pivot_table(index='RA', columns='Ano Pesquisa', values=['IEG', 'IDA'])

# Calcular Delta IEG de 2022 para 2023
df_pivot['Delta_IEG_22_23'] = df_pivot[('IEG', 2023)] - df_pivot[('IEG', 2022)]

# Calcular Delta IDA de 2023 para 2024
df_pivot['Delta_IDA_23_24'] = df_pivot[('IDA', 2024)] - df_pivot[('IDA', 2023)]

# Selecionar apenas as colunas de interesse para a correlação
df_correlation = df_pivot[['Delta_IEG_22_23', 'Delta_IDA_23_24']].dropna()

# Calcular a correlação entre Delta_IEG_22_23 e Delta_IDA_23_24
correlation_delta = df_correlation['Delta_IEG_22_23'].corr(df_correlation['Delta_IDA_23_24'])

print(f"Correlação entre Delta IEG (2022-2023) e Delta IDA (2023-2024): {correlation_delta:.3f}")

```

### Análise da Coerência entre IAA, IDA e IEG

Para verificar a coerência entre a Autoavaliação (IAA), o Desempenho Real (IDA) e o Engajamento (IEG), vamos analisar as correlações entre essas variáveis, além de explorar a distribuição conjunta em diferentes cenários.

```python
# Definir as categorias de autoavaliação para IAA
def categorize_iaa(iaa):
    if pd.isna(iaa):
        return pd.NA
    if iaa < 5.0:
        return 'Baixa Autoavaliação'
    elif 5.0 <= iaa < 7.5:
        return 'Média Autoavaliação'
    else:
        return 'Alta Autoavaliação'

# Definir as categorias de desempenho para IDA (repetido para garantir definição no escopo da célula)
def categorize_ida(ida):
    if pd.isna(ida):
        return pd.NA
    if ida < 5.0:
        return 'Baixo Desempenho'
    elif 5.0 <= ida < 7.5:
        return 'Médio Desempenho'
    else:
        return 'Alto Desempenho'

# Definir as categorias de engajamento para IEG (repetido para garantir definição no escopo da célula)
def categorize_ieg(ieg):
    if pd.isna(ieg):
        return pd.NA
    if ieg < 5.0:
        return 'Baixo Engajamento'
    elif 5.0 <= ieg < 7.5:
        return 'Médio Engajamento'
    else:
        return 'Alto Engajamento'

df_long['IAA_Categoria'] = df_long['IAA'].apply(categorize_iaa)
df_long['IDA_Categoria'] = df_long['IDA'].apply(categorize_ida)
df_long['IEG_Categoria'] = df_long['IEG'].apply(categorize_ieg)

# Tabela de contingência: IAA_Categoria vs IDA_Categoria
print("Tabela de Contingência: IAA_Categoria vs IDA_Categoria:")
display(pd.crosstab(df_long['IAA_Categoria'].fillna('Não Aplicável'), df_long['IDA_Categoria'].fillna('Não Aplicável'), dropna=False))

# Tabela de contingência: IAA_Categoria vs IEG_Categoria
print("\nTabela de Contingência: IAA_Categoria vs IEG_Categoria:")
display(pd.crosstab(df_long['IAA_Categoria'].fillna('Não Aplicável'), df_long['IEG_Categoria'].fillna('Não Aplicável'), dropna=False))

# Tabela de contingência: IAA_Categoria vs IDA_Categoria por Ano Pesquisa (iterativo)
print("\nTabela de Contingência por Ano de Pesquisa: IAA_Categoria vs IDA_Categoria")
for ano, group_df in df_long.groupby('Ano Pesquisa'):
    print(f"\nAno Pesquisa: {ano}")
    if 'IAA_Categoria' in group_df.columns and 'IDA_Categoria' in group_df.columns:
        try:
            crosstab_result = pd.crosstab(group_df['IAA_Categoria'].fillna('Não Aplicável'), group_df['IDA_Categoria'].fillna('Não Aplicável'), dropna=False)
            if crosstab_result.empty:
                print(f"Tabela de contingência vazia para Ano Pesquisa {ano}")
            else:
                print(crosstab_result.to_string()) # Use to_string to force text output and avoid display truncation
        except Exception as e:
            print(f"Erro ao gerar tabela de contingência para Ano Pesquisa {ano}: {e}")
    else:
        print(f"Colunas 'IAA_Categoria' ou 'IDA_Categoria' ausentes para o ano {ano}")

# Tabela de contingência: IAA_Categoria vs IEG_Categoria por Ano Pesquisa (iterativo)
print("\nTabela de Contingência por Ano de Pesquisa: IAA_Categoria vs IEG_Categoria")
for ano, group_df in df_long.groupby('Ano Pesquisa'):
    print(f"\nAno Pesquisa: {ano}")
    if 'IAA_Categoria' in group_df.columns and 'IEG_Categoria' in group_df.columns:
        try:
            crosstab_result = pd.crosstab(group_df['IAA_Categoria'].fillna('Não Aplicável'), group_df['IEG_Categoria'].fillna('Não Aplicável'), dropna=False)
            if crosstab_result.empty:
                print(f"Tabela de contingência vazia para Ano Pesquisa {ano}")
            else:
                print(crosstab_result.to_string()) # Use to_string to force text output and avoid display truncation
        except Exception as e:
            print(f"Erro ao gerar tabela de contingência para Ano Pesquisa {ano}: {e}")
    else:
        print(f"Colunas 'IAA_Categoria' ou 'IEG_Categoria' ausentes para o ano {ano}")
```

### Síndrome do Impostor vs. Excesso de Confiança

Vamos calcular um "Índice de Realidade" (IAA - IDA) para cada aluno. Com base nesse índice, podemos categorizar os alunos da seguinte forma:

*   **Síndrome do Impostor**: IAA < IDA (o aluno subestima sua capacidade)
*   **Excesso de Confiança**: IAA > IDA (o aluno superestima sua capacidade)
*   **Coerente**: IAA = IDA ou a diferença é mínima (percepção alinhada com o desempenho)

Analisaremos qual perfil é predominante na Passos Mágicos.

```python
# Calcular o 'Índice de Realidade' (IAA - IDA)
df_long['Indice_Realidade'] = df_long['IAA'] - df_long['IDA']

# Categorizar o perfil
def categorize_reality_index(index):
    if pd.isna(index):
        return pd.NA
    if index < -0.5: # Considerar uma pequena margem para 'coerente'
        return 'Síndrome do Impostor'
    elif index > 0.5: # Considerar uma pequena margem para 'coerente'
        return 'Excesso de Confiança'
    else:
        return 'Coerente'

df_long['Perfil_Autoavaliacao'] = df_long['Indice_Realidade'].apply(categorize_reality_index)

# Contar a predominância dos perfis
perfil_counts = df_long['Perfil_Autoavaliacao'].value_counts(dropna=False)
print("Predominância dos Perfis de Autoavaliação:")
display(perfil_counts)

# Calcular a proporção dos perfis por ano
perfil_por_ano = df_long.groupby('Ano Pesquisa')['Perfil_Autoavaliacao'].value_counts(normalize=True).unstack(fill_value=0)
print("\nProporção dos Perfis de Autoavaliação por Ano:")
display(perfil_por_ano)

# Visualizar a proporção dos perfis por ano
fig_perfil = px.bar(perfil_por_ano.reset_index().melt(id_vars='Ano Pesquisa'),
                     x='Ano Pesquisa', y='value', color='Perfil_Autoavaliacao',
                     title='Proporção de Perfis de Autoavaliação por Ano',
                     labels={'value': 'Proporção', 'Perfil_Autoavaliacao': 'Perfil'},
                     barmode='stack')
fig_perfil.update_layout(yaxis_tickformat='.0%')
fig_perfil.update_xaxes(dtick=1)
fig_perfil.show()
```

*   **Alto IPS**: IPS >= 7.5

## Padrões Psicossociais (IPS) que Antecedem Quedas de Desempenho (IDA) ou Engajamento (IEG)

Para determinar se os padrões psicossociais (IPS) antecedem quedas no desempenho acadêmico (IDA) ou no engajamento (IEG), realizaremos as seguintes análises:

1.  **Correlações Defasadas (Lagged Correlations)**: Calcularemos a correlação entre o IPS de um ano (N) e o IDA/IEG do ano seguinte (N+1). Uma correlação positiva significativa indicaria que um IPS mais alto em um ano pode estar associado a um IDA/IEG mais alto no ano seguinte.
2.  **Análise de IPS antes de Quedas**: Identificaremos os alunos que tiveram uma queda significativa no IDA ou IEG entre um ano e o próximo e, em seguida, analisaremos seus valores de IPS no ano anterior à queda.

```python
# 1. Correlações Defasadas (Lagged Correlations)

# Criar um DataFrame com RA, Ano Pesquisa, IPS, IDA e IEG
df_lag = df_long[['RA', 'Ano Pesquisa', 'IPS', 'IDA', 'IEG']].copy()
df_lag.sort_values(by=['RA', 'Ano Pesquisa'], inplace=True)

# Calcular o IDA e IEG do ano seguinte
df_lag['IDA_next_year'] = df_lag.groupby('RA')['IDA'].shift(-1)
df_lag['IEG_next_year'] = df_lag.groupby('RA')['IEG'].shift(-1)

# Calcular as correlações defasadas
# Correlação entre IPS no ano N e IDA no ano N+1
correlation_ips_ida_lagged = df_lag['IPS'].corr(df_lag['IDA_next_year'])
print(f"Correlação entre IPS (Ano N) e IDA (Ano N+1): {correlation_ips_ida_lagged:.3f}")

# Correlação entre IPS no ano N e IEG no ano N+1
correlation_ips_ieg_lagged = df_lag['IPS'].corr(df_lag['IEG_next_year'])
print(f"Correlação entre IPS (Ano N) e IEG (Ano N+1): {correlation_ips_ieg_lagged:.3f}")

print("\nIsolando para cada ano de IPS e o ano subsequente de IDA/IEG:")
corr_2022_2023_ida = None
corr_2022_2023_ieg = None
corr_2023_2024_ida = None
corr_2023_2024_ieg = None

for year in sorted(df_lag['Ano Pesquisa'].unique()):
    if year + 1 in df_lag['Ano Pesquisa'].unique(): # Ensure there's a next year to compare
        df_current_year = df_lag[df_lag['Ano Pesquisa'] == year]
        df_next_year = df_lag[df_lag['Ano Pesquisa'] == year + 1]

        # Merge para alinhar IPS do ano N com IDA/IEG do ano N+1 para os mesmos RAs
        merged_df = pd.merge(df_current_year[['RA', 'IPS']],
                             df_next_year[['RA', 'IDA', 'IEG']],
                             on='RA',
                             how='inner', # Use inner to ensure only RAs present in both years are considered
                             suffixes=('_N', '_N_plus_1'))

        if not merged_df.empty:
            corr_ida = merged_df['IPS'].corr(merged_df['IDA'])
            corr_ieg = merged_df['IPS'].corr(merged_df['IEG'])
            print(f"  Ano {year} (IPS) -> Ano {year+1} (IDA): {corr_ida:.3f}")
            print(f"  Ano {year} (IPS) -> Ano {year+1} (IEG): {corr_ieg:.3f}")

            if year == 2022:
                corr_2022_2023_ida = corr_ida
                corr_2022_2023_ieg = corr_ieg
            elif year == 2023:
                corr_2023_2024_ida = corr_ida
                corr_2023_2024_ieg = corr_ieg
```

```python
print(f"Correlação entre IPS (Ano N) e IDA (Ano N+1) geral: {correlation_ips_ida_lagged:.3f}")
print(f"Correlação entre IPS (Ano N) e IEG (Ano N+1) geral: {correlation_ips_ieg_lagged:.3f}")

if corr_2022_2023_ida is not None: print(f"Correlação entre IPS (2022) e IDA (2023): {corr_2022_2023_ida:.3f}")
if corr_2022_2023_ieg is not None: print(f"Correlação entre IPS (2022) e IEG (2023): {corr_2022_2023_ieg:.3f}")
if corr_2023_2024_ida is not None: print(f"Correlação entre IPS (2023) e IDA (2024): {corr_2023_2024_ida:.3f}")
if corr_2023_2024_ieg is not None: print(f"Correlação entre IPS (2023) e IEG (2024): {corr_2023_2024_ieg:.3f}")
```

### Análise de IPS Anterior a Quedas no IDA e IEG

```python
import pandas as pd
import numpy as np
import plotly.express as px

# 2. Análise de IPS antes de Quedas

# Definir o que consideramos uma 'queda significativa'
# Por exemplo, uma queda de 1.0 ponto ou mais
drop_threshold_ida = -1.0
drop_threshold_ieg = -1.0

# Identify drops in IDA
df_ida_drops = df_evolution_ida[df_evolution_ida['Mudanca_IDA'] < drop_threshold_ida].copy()
# Add a column for the year *prior* to the drop year
df_ida_drops['Ano_IPS_anterior'] = df_ida_drops['Ano Pesquisa'] - 1

# Get IPS from the year PRIOR to the IDA drop
merged_ida_ips_prev = pd.merge(
    df_ida_drops[['RA', 'Ano Pesquisa', 'Ano_IPS_anterior']],  # Left side contains RA, drop_year, prior_ips_year
    df_long[['RA', 'Ano Pesquisa', 'IPS']],  # Right side contains RA, actual_ips_year, IPS_value
    left_on=['RA', 'Ano_IPS_anterior'],  # Join on RA and match prior_ips_year with actual_ips_year
    right_on=['RA', 'Ano Pesquisa'],
    how='left',
    suffixes=('_drop_context', '_ips_prev_year') # Add suffixes to avoid dropping original 'Ano Pesquisa'
)

# The IPS value from df_long will now be in a column named 'IPS'. We need to rename it.
# The 'Ano Pesquisa' from the right side will be 'Ano Pesquisa_ips_prev_year'.
# The 'Ano Pesquisa' from the left side (drop context) is 'Ano Pesquisa_drop_context'.

# Select relevant columns and rename IPS
df_ida_drops_ips_prev = merged_ida_ips_prev[[
    'RA',
    'Ano Pesquisa_drop_context', # The year of the drop
    'IPS'                       # The IPS value from the previous year
]].rename(columns={
    'Ano Pesquisa_drop_context': 'Ano Pesquisa',
    'IPS': 'IPS_ano_anterior_a_queda'
})

# Drop rows where previous IPS was not found
df_ida_drops_ips_prev.dropna(subset=['IPS_ano_anterior_a_queda'], inplace=True)


# Identify drops in IEG
df_ieg_drops = df_evolution_ieg[df_evolution_ieg['Mudanca_IEG'] < drop_threshold_ieg].copy()
df_ieg_drops['Ano_IPS_anterior'] = df_ieg_drops['Ano Pesquisa'] - 1

# Get IPS from the year PRIOR to the IEG drop
merged_ieg_ips_prev = pd.merge(
    df_ieg_drops[['RA', 'Ano Pesquisa', 'Ano_IPS_anterior']],
    df_long[['RA', 'Ano Pesquisa', 'IPS']],
    left_on=['RA', 'Ano_IPS_anterior'],
    right_on=['RA', 'Ano Pesquisa'],
    how='left',
    suffixes=('_drop_context', '_ips_prev_year')
)

df_ieg_drops_ips_prev = merged_ieg_ips_prev[[
    'RA',
    'Ano Pesquisa_drop_context', # The year of the drop
    'IPS'                       # The IPS value from the previous year
]].rename(columns={
    'Ano Pesquisa_drop_context': 'Ano Pesquisa',
    'IPS': 'IPS_ano_anterior_a_queda'
})

df_ieg_drops_ips_prev.dropna(subset=['IPS_ano_anterior_a_queda'], inplace=True)


# Calcular a média do IPS antes das quedas
# Handle cases where no drops are found or no prior IPS data is available for means calculation
mean_ips_before_ida_drop = np.nan
if not df_ida_drops_ips_prev.empty:
    mean_ips_before_ida_drop = df_ida_drops_ips_prev['IPS_ano_anterior_a_queda'].mean()

mean_ips_before_ieg_drop = np.nan
if not df_ieg_drops_ips_prev.empty:
    mean_ips_before_ieg_drop = df_ieg_drops_ips_prev['IPS_ano_anterior_a_queda'].mean()

# Média geral do IPS para comparação
mean_ips_overall = df_long['IPS'].mean()

print(f"Média do IPS no ano anterior a quedas significativas de IDA: {mean_ips_before_ida_drop:.3f}")
print(f"Média do IPS no ano anterior a quedas significativas de IEG: {mean_ips_before_ieg_drop:.3f}")
print(f"Média geral do IPS (todos os anos e alunos): {mean_ips_overall:.3f}")

# Visualizar a distribuição do IPS antes das quedas vs. distribuição geral do IPS

# Criar um DataFrame para as distribuições
df_ips_dist = pd.DataFrame({
    'IPS': df_long['IPS'],
    'Tipo': 'IPS Geral'
})

# Append IPS before IDA drops
if not df_ida_drops_ips_prev.empty:
    df_ips_dist = pd.concat([
        df_ips_dist,
        pd.DataFrame({'IPS': df_ida_drops_ips_prev['IPS_ano_anterior_a_queda'], 'Tipo': 'IPS antes de Queda IDA'})
    ])
# Append IPS before IEG drops
if not df_ieg_drops_ips_prev.empty:
    df_ips_dist = pd.concat([
        df_ips_dist,
        pd.DataFrame({'IPS': df_ieg_drops_ips_prev['IPS_ano_anterior_a_queda'], 'Tipo': 'IPS antes de Queda IEG'})
    ])

# Drop NaN values from the concatenated DataFrame for plotting
df_ips_dist.dropna(subset=['IPS'], inplace=True)

fig_ips_drops = px.histogram(df_ips_dist, x='IPS', color='Tipo', barmode='overlay',
                             title='Distribuição do IPS antes de Quedas de IDA/IEG vs. IPS Geral',
                             labels={'IPS': 'Valor do IPS', 'Tipo': 'Contexto do IPS'},
                             opacity=0.7)
fig_ips_drops.update_layout(xaxis_title='Valor do IPS', yaxis_title='Contagem')
fig_ips_drops.show()
```

### Análise da "Bóia de Salvação" (IPV): IPS, IDA e o papel do IPV

Vamos investigar se um alto IPV (Indicador de Ponto de Virada) atua como um amortecedor para o desempenho acadêmico (IDA) quando o Indicador Psicossocial (IPS) do aluno sofre uma queda. A hipótese é que, mesmo com problemas psicossociais (queda de IPS), um aluno com alto IPV pode conseguir sustentar ou ter uma queda menor no IDA.

```python
# Definir o que consideramos uma 'queda significativa' no IPS
drop_threshold_ips = -1.0 # Ex: queda de 1.0 ponto ou mais

# Preparar o DataFrame para análise
df_salvacao = df_long[['RA', 'Ano Pesquisa', 'IPS', 'IDA', 'IPV']].copy()
df_salvacao.sort_values(by=['RA', 'Ano Pesquisa'], inplace=True)

# Calcular a mudança no IPS e IDA entre anos consecutivos
df_salvacao['IPS_anterior'] = df_salvacao.groupby('RA')['IPS'].shift(1)
df_salvacao['Mudanca_IPS'] = df_salvacao['IPS'] - df_salvacao['IPS_anterior']

df_salvacao['IDA_anterior'] = df_salvacao.groupby('RA')['IDA'].shift(1)
df_salvacao['Mudanca_IDA'] = df_salvacao['IDA'] - df_salvacao['IDA_anterior']

# Identificar alunos com queda significativa no IPS
df_ips_queda = df_salvacao[df_salvacao['Mudanca_IPS'] < drop_threshold_ips].copy()

# Para estes alunos, vamos pegar o IDA do ano seguinte à queda do IPS e o IPV do ano da queda do IPS
# Queremos ver se o IDA no ano 'N+1' é menos afetado se o IPV no ano 'N' (ano da queda do IPS) é alto.

# Trazer o IDA do ano subsequente à queda do IPS (IDA_N+1)
df_ips_queda['IDA_ano_seguinte_a_queda'] = df_ips_queda.groupby('RA')['IDA'].shift(-1)

# Trazer o IPV do ano da queda do IPS (IPV_N)
df_ips_queda['IPV_no_ano_da_queda_IPS'] = df_ips_queda['IPV']

# Filtrar para ter apenas as linhas onde houve a queda de IPS e temos o IDA do ano seguinte
df_analise_salvacao = df_ips_queda.dropna(subset=['Mudanca_IPS', 'IDA_ano_seguinte_a_queda', 'IPV_no_ano_da_queda_IPS']).copy()

# Exibir as primeiras linhas da análise de salvação
print("DataFrame para Análise de 'Bóia de Salvação' (IPS, IDA, IPV):")
display(df_analise_salvacao.head())

# Agora, vamos categorizar o IPV para ver o impacto
def categorize_ipv(ipv):
    if pd.isna(ipv):
        return pd.NA
    if ipv >= 7.5: # Usar 7.5 como limite para 'alto IPV'
        return 'Alto IPV'
    else:
        return 'Baixo/Médio IPV'

df_analise_salvacao['IPV_Categoria'] = df_analise_salvacao['IPV_no_ano_da_queda_IPS'].apply(categorize_ipv)

# Analisar o IDA_ano_seguinte_a_queda em função da IPV_Categoria
print("\nMédia da Mudança do IDA após Queda do IPS, por Categoria de IPV:")
display(df_analise_salvacao.groupby('IPV_Categoria')['IDA_ano_seguinte_a_queda'].mean().reset_index())

# Podemos também analisar a distribuição da 'Mudanca_IDA' para esses grupos
print("\nMédia da Mudança do IDA (no ano da queda do IPS) por Categoria de IPV:")
display(df_analise_salvacao.groupby('IPV_Categoria')['Mudanca_IDA'].mean().reset_index())

# Visualização: Box plot da Mudança do IDA, agrupado por Categoria de IPV
import plotly.express as px

fig = px.box(df_analise_salvacao, x='IPV_Categoria', y='IDA_ano_seguinte_a_queda',
             title='IDA no Ano Seguinte à Queda do IPS por Categoria de IPV',
             labels={'IDA_ano_seguinte_a_queda': 'IDA no Ano Seguinte', 'IPV_Categoria': 'Categoria de IPV (Ano da Queda do IPS)'},
             color='IPV_Categoria')
fig.show()

fig2 = px.box(df_analise_salvacao, x='IPV_Categoria', y='Mudanca_IDA',
             title='Mudança no IDA (no ano da queda do IPS) por Categoria de IPV',
             labels={'Mudanca_IDA': 'Mudança no IDA', 'IPV_Categoria': 'Categoria de IPV (Ano da Queda do IPS)'},
             color='IPV_Categoria')
fig2.show()
```

### Validação Clínica: Cruzamento de Quedas de IPS com Recomendações de Psicologia

Agora, vamos verificar se grandes quedas no IPS estão alinhadas com as recomendações da equipe de psicologia. Para isso, iremos:

1.  Identificar os alunos que tiveram uma queda significativa no IPS (já feito na análise anterior).

# VI. Aspectos psicopedagógicos (IPP)

As avaliações psicopedagógicas (IPP) confirmam ou contradizem a defasagem identificada pelo IAN?

```python
from scipy import stats

# marca o ano
df_2023['Ano'] = 2023
df_2024['Ano'] = 2024

# limpa números com vírgula decimal
def clean_num(s):
    return pd.to_numeric(s.astype(str).str.replace(',', '.'), errors='coerce')

for df in [df_2023, df_2024]:
    df['IAN_n'] = clean_num(df['IAN'])
    df['IPP_n'] = clean_num(df['IPP'])

# combina 2023 + 2024 (2022 não tem IPP)
comb = pd.concat([df_2023, df_2024], ignore_index=True)
comb = comb.dropna(subset=['IAN_n', 'IPP_n'])

# média de IPP por nível de IAN
resumo = comb.groupby('IAN_n')['IPP_n'].agg(['count', 'mean', 'median', 'std'])
print(resumo)

# teste estatístico simples
grupos = [g['IPP_n'].values for _, g in comb.groupby('IAN_n')]
h, p = stats.kruskal(*grupos)
print(f'\nKruskal-Wallis: H={h:.3f}, p={p:.5f}')

rho, p_rho = stats.spearmanr(comb['IAN_n'], comb['IPP_n'])
print(f'Spearman IAN x IPP: rho={rho:.3f}, p={p_rho:.5f}')
```

###Gráfico 6. Aspectos psicopedagógicos (IPP)

```python
import matplotlib.pyplot as plt

# reaproveita o 'comb' que você já montou para a questão 6 (2023+2024, com IAN_n e IPP_n)
resumo6 = comb.groupby('IAN_n')['IPP_n'].mean()

labels = ['Severa (2,5)', 'Moderada (5,0)', 'Sem defasagem (10,0)']
valores = resumo6.values

plt.figure(figsize=(7,5))
barras = plt.bar(labels, valores, color=['#c0392b', '#e6a817', '#2e8b57'])
plt.ylim(0, 10)
plt.ylabel('IPP médio')
plt.title('IPP médio por nível de defasagem (IAN) — 2023 e 2024')

# coloca o valor em cima de cada barra
for b, v in zip(barras, valores):
    plt.text(b.get_x() + b.get_width()/2, v + 0.15, f'{v:.2f}', ha='center', fontweight='bold')

plt.tight_layout()
plt.show()
```

### Resposta 6. Aspectos psicopedagógicos (IPP):
(apenas 2023 e 2024 pois 2022 não apresenta IPP)

Confirmam parcialmente — na direção certa, mas com relação fraca.

### O que os números mostram:

Nível de IAN | Alunos (N) | IPP médio | IPP mediana
2,5 — defasagem severa | 17 | 7,01 | 7,50
5,0 — defasagem moderada | 1.067 | 7,46 | 7,50
10,0 — sem defasagem | 908 | 7,68 | 7,71

A tendência é a esperada: quanto maior a defasagem (IAN menor), menor o IPP médio. Isso já é um primeiro sinal de coerência entre os dois indicadores.

### O que os testes confirmam:

Kruskal-Wallis (H=34,7, p<0,00001): existe, sim, diferença estatisticamente real entre os três grupos — não é coincidência ou ruído dos dados.
Spearman (rho=0,131, p<0,00001): a relação segue a direção esperada (mais defasagem → IPP mais baixo), mas é fraca. Um rho de 0,13 está bem longe de indicar uma relação forte — está mais para "existe uma leve tendência" do que para "um indicador espelha o outro".

### Como interpretar isso na prática:

Olhando as médias, a diferença entre o grupo com defasagem severa (7,01) e o grupo sem defasagem (7,68) é de menos de 1 ponto numa escala de 0 a 10. Ou seja, mesmo alunos com defasagem acadêmica severa continuam recebendo avaliações psicopedagógicas relativamente boas na maioria dos casos.

### Conclusão para a apresentação:

O IPP não contradiz o IAN — ele caminha na mesma direção. Mas também não confirma fortemente a defasagem — ele não funciona como um espelho do atraso acadêmico. Isso sugere que os dois indicadores medem dimensões parcialmente diferentes do aluno: o IAN capta a defasagem de fase/série, enquanto o IPP capta aspectos psicopedagógicos (comportamento, adaptação, processo de aprendizagem) que nem sempre andam junto com o atraso formal.

Também vale mencionar a ressalva estatística: o grupo de defasagem severa tem só 17 alunos (nos dois anos combinados), então essa conclusão específica desse grupo tem menos robustez que as dos outros dois.

#VII. Ponto de virada (IPV)

Quais comportamentos - acadêmicos, emocionais ou de engajamento - mais influenciam o IPV ao longo do tempo?

```python
import pandas as pd
import numpy as np

def clean_num(s):
    return pd.to_numeric(s.astype(str).str.replace(',', '.'), errors='coerce')

for df in [df_2022, df_2023, df_2024]:
    for c in ['IPV','IDA','IAA','IEG','IPS']:
        df[c+'_n'] = clean_num(df[c])
df_2023['IPP_n'] = clean_num(df_2023['IPP'])
df_2024['IPP_n'] = clean_num(df_2024['IPP'])

def reg(df, preds, ano):
    sub = df[preds + ['IPV_n']].dropna()
    X = sub[preds].values
    Xz = (X - X.mean(axis=0)) / X.std(axis=0)
    y = sub['IPV_n'].values
    yz = (y - y.mean()) / y.std()
    Xz1 = np.column_stack([np.ones(len(Xz)), Xz])
    coefs, _, _, _ = np.linalg.lstsq(Xz1, yz, rcond=None)
    y_pred = Xz1 @ coefs
    r2 = 1 - np.sum((yz - y_pred)**2) / np.sum((yz - yz.mean())**2)
    print(f'--- {ano} (N={len(sub)}) R2={r2:.3f} ---')
    for nome, c in zip(['const'] + preds, coefs):
        print(f'  {nome}: {c:.4f}')
    return dict(zip(preds, coefs[1:]))

preds_base = ['IDA_n', 'IAA_n', 'IEG_n', 'IPS_n']
p22 = reg(df_2022, preds_base, 2022)
p23 = reg(df_2023, preds_base + ['IPP_n'], 2023)
p24 = reg(df_2024, preds_base + ['IPP_n'], 2024)
```

1) O que é estável ao longo do tempo: em todos os 3 anos, acadêmico (IDA) e engajamento (IEG) são disparadamente os dois comportamentos que mais empurram o IPV — os dois sempre aparecem muito acima dos demais. Já o emocional (IAA) e o psicossocial (IPS) praticamente não influenciam o IPV em nenhum ano — os coeficientes ficam perto de zero o tempo todo, alguns até levemente negativos. Ou seja: não é a autopercepção do aluno nem questões psicossociais que empurram o ponto de virada — é o que ele entrega academicamente e o quanto se engaja.

2) O que muda ao longo do tempo (o achado mais interessante): quando trazemos o IPP (psicopedagógico) pra jogo em 2023 e 2024, ele cresce de forma expressiva ano a ano — de peso 0,355 em 2023 para 0,613 em 2024, se tornando o fator isolado mais importante, ultrapassando inclusive o acadêmico. Ao mesmo tempo, o peso do IDA vai caindo (0,40 → 0,31 → 0,16 ao longo dos 3 anos). O modelo também fica muito mais explicativo quando o IPP entra: o R² salta de 0,46 (2023) para 0,65 em 2024.

### Conclusão para a apresentação:

O comportamento acadêmico e o engajamento seguem sendo os pilares "clássicos" que sustentam o ponto de virada, mas há uma tendência clara de que o acompanhamento psicopedagógico (IPP) está se tornando o fator mais decisivo para um aluno atingir o ponto de virada, superando inclusive o desempenho acadêmico em 2024. Isso é um insight forte pra Passos Mágicos: sugere que investir no acompanhamento psicopedagógico individualizado pode ter hoje mais impacto no ponto de virada do que apenas reforçar conteúdo acadêmico — o aspecto emocional/psicossocial isolado (IAA, IPS), por outro lado, não parece ser o caminho.

```python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler

def clean_num(s):
    return pd.to_numeric(s.astype(str).str.replace(',', '.'), errors='coerce')

for df in [df_2022, df_2023, df_2024]:
    for c in ['IPV','IDA','IAA','IEG','IPS','IAN']:
        df[c+'_n'] = clean_num(df[c])
df_2023['IPP_n'] = clean_num(df_2023['IPP'])
df_2024['IPP_n'] = clean_num(df_2024['IPP'])

def reg_std(sub, preds, target='IPV_n'):
    X = sub[preds].values
    Xz = (X - X.mean(axis=0)) / X.std(axis=0)
    y = sub[target].values
    yz = (y - y.mean()) / y.std()
    Xz1 = np.column_stack([np.ones(len(Xz)), Xz])
    coefs, _, _, _ = np.linalg.lstsq(Xz1, yz, rcond=None)
    y_pred = Xz1 @ coefs
    r2 = 1 - np.sum((yz-y_pred)**2) / np.sum((yz-yz.mean())**2)
    return coefs, r2
```

### Conclusão para a apresentação:

Para elevar o INDE, a combinação mais poderosa é reforçar simultaneamente desempenho acadêmico (IDA) e engajamento (IEG) — eles se retroalimentam e concentram a maior parte do peso. ##Setup para as análises complementares questão 8

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

def clean_num(s):
    return pd.to_numeric(s.astype(str).str.replace(',', '.'), errors='coerce')

df_2023['INDE_n'] = clean_num(df_2023['INDE'])
df_2024['INDE_n'] = clean_num(df_2024['INDE'])
for df in [df_2023, df_2024]:
    for c in ['IDA','IEG','IPS','IPP','IAN','IPV']:
        df[c+'_n'] = clean_num(df[c])

def reg_std(sub, preds, target='INDE_n'):
    X = sub[preds].values
    Xz = (X - X.mean(axis=0)) / X.std(axis=0)
    y = sub[target].values
    yz = (y - y.mean()) / y.std()
    Xz1 = np.column_stack([np.ones(len(Xz)), Xz])
    coefs, _, _, _ = np.linalg.lstsq(Xz1, yz, rcond=None)
    y_pred = Xz1 @ coefs
    r2 = 1 - np.sum((yz-y_pred)**2) / np.sum((yz-yz.mean())**2)
    return coefs, r2

cols = ['IDA_n','IEG_n','IPS_n','IPP_n','IAN_n','IPV_n','INDE_n']
comb8 = pd.concat([df_2023[cols], df_2024[cols]], ignore_index=True).dropna()
```

##Análise 1 — Existe sinergia entre IDA e IEG, ou é só soma das partes?

```python
sub1 = comb8.copy()
sub1['INT'] = sub1['IDA_n'] * sub1['IEG_n']
coefs, r2 = reg_std(sub1, ['IDA_n','IEG_n','INT'])
resultado1 = {'IDA': coefs[1], 'IEG': coefs[2], 'INT': coefs[3]}
print(f'N={len(sub1)} R2={r2:.3f} | IDA={coefs[1]:.3f} IEG={coefs[2]:.3f} INT={coefs[3]:.3f}')
```

###Gráfico análise 1

```python
nomes = ['IDA', 'IEG', 'IDA × IEG (interação)']
valores = [resultado1['IDA'], resultado1['IEG'], resultado1['INT']]

plt.figure(figsize=(7,5))
barras = plt.bar(nomes, valores, color=['#2e6f95','#e6a817','#c0392b'])
plt.ylabel('Peso padronizado sobre o INDE')
plt.title('O efeito combinado de IDA e IEG é maior que a soma das partes?')
for b, v in zip(barras, valores):
    plt.text(b.get_x()+b.get_width()/2, v + 0.01, f'{v:.3f}', ha='center', fontweight='bold')
plt.tight_layout()
plt.show()
```

##Análise 2 — O que diferencia o top 10% do bottom 10% em INDE?

```python
p90 = comb8['INDE_n'].quantile(0.90)
p10 = comb8['INDE_n'].quantile(0.10)
top = comb8[comb8['INDE_n'] >= p90]
bot = comb8[comb8['INDE_n'] <= p10]

resultado2 = {}
print(f'Top10% N={len(top)} (INDE>={p90:.2f}) | Bottom10% N={len(bot)} (INDE<={p10:.2f})')
for var in ['IDA_n','IEG_n','IPS_n','IPP_n']:
    diff = top[var].mean() - bot[var].mean()
    resultado2[var] = diff
    print(f'{var}: top={top[var].mean():.2f} bottom={bot[var].mean():.2f} diff={diff:.2f}')
```

###Gráfico análise 2

```python
nomes = ['IDA', 'IEG', 'IPS', 'IPP']
valores = [resultado2['IDA_n'], resultado2['IEG_n'], resultado2['IPS_n'], resultado2['IPP_n']]

plt.figure(figsize=(7,5))
barras = plt.bar(nomes, valores, color=['#2e6f95','#e6a817','#7f8c8d','#c0392b'])
plt.ylabel('Diferença de média (Top 10% − Bottom 10%)')
plt.title('O que mais separa os melhores dos piores alunos em INDE?')
for b, v in zip(barras, valores):
    plt.text(b.get_x()+b.get_width()/2, v + 0.05, f'{v:.2f}', ha='center', fontweight='bold')
plt.tight_layout()
plt.show()
```

##Análise 3 — A combinação ideal muda entre 2023 e 2024?

```python
resultado3 = {}
for df, ano in [(df_2023, 2023), (df_2024, 2024)]:
    sub = df[['IDA_n','IEG_n','IPS_n','IPP_n','INDE_n']].dropna()
    coefs, r2 = reg_std(sub, ['IDA_n','IEG_n','IPS_n','IPP_n'])
    resultado3[ano] = {'IDA': coefs[1], 'IEG': coefs[2], 'IPS': coefs[3], 'IPP': coefs[4], 'R2': r2}
    print(f'{ano}: N={len(sub)} R2={r2:.3f} | IDA={coefs[1]:.3f} IEG={coefs[2]:.3f} IPS={coefs[3]:.3f} IPP={coefs[4]:.3f}')
```
